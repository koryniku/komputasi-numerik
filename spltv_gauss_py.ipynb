```python
import numpy as np

def selesaikan_spltv(A, b):
    """
    Menyelesaikan Sistem Persamaan Linear Tiga Variabel (SPLTV)
    menggunakan metode Eliminasi Gauss.
    A: Matriks koefisien (3x3)
    b: Vektor konstanta (3x1)
    """
    # Menggabungkan matriks A dan b menjadi augmented matrix
    n = len(b)
    augmented_mat = np.hstack((A, b.reshape(-1, 1)))

    print("--- Matriks Awal (Augmented Matrix) ---")
    print(augmented_mat)
    print("-" * 40)

    # Proses Eliminasi Gauss (Forward Elimination)
    for i in range(n):
        # Pivoting partial (opsional untuk keamanan numerik)
        max_row = i + np.argmax(np.abs(augmented_mat[i:, i]))
        if augmented_mat[max_row, i] == 0:
            raise ValueError("Sistem persamaan tidak memiliki solusi unik (determinan nol).")

        # Tukar baris jika diperlukan
        if max_row != i:
            augmented_mat[[i, max_row]] = augmented_mat[[max_row, i]]

        # Membuat elemen di bawah pivot menjadi 0
        for j in range(i + 1, n):
            faktor = augmented_mat[j, i] / augmented_mat[i, i]
            augmented_mat[j, i:] -= faktor * augmented_mat[i, i:]

    print("--- Matriks Setelah Eliminasi Gauss ---")
    print(np.round(augmented_mat, 4))
    print("-" * 40)

    # Substitusi Balik (Back Substitution)
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (augmented_mat[i, -1] - np.dot(augmented_mat[i, i+1:n], x[i+1:n])) / augmented_mat[i, i]

    return x

if __name__ == "__main__":
    # Contoh Kasus SPLTV:
    # 2x + y + z = 4
    # x + 3y + 2z = 6
    # x + y + 3z = 7

    # Matriks Koefisien A
    A = np.array([
        [2.0, 1.0, 1.0],
        [1.0, 3.0, 2.0],
        [1.0, 1.0, 3.0]
    ])

    # Vektor Konstanta b
    b = np.array([4.0, 6.0, 7.0])

    try:
        solusi = selesaikan_spltv(A, b)
        print("Hasil Penyelesaian:")
        print(f"x = {solusi[0]:.4f}")
        print(f"y = {solusi[1]:.4f}")
        print(f"z = {solusi[2]:.4f}")
    except Exception as e:
        print(f"Terjadi kesalahan: {e}")
```
